```python
# Smart Library Desk - Simple Reflex Agent

THRESHOLD = 60  # Noise limit (in dB)

print("=== Smart Library Desk ===")

while True:
    noise = int(input("\nEnter current noise level (dB): "))

    if noise > THRESHOLD:
        print("🔴 RED LED: ON")
        print("📢 PLEASE MAINTAIN SILENCE")
    else:
        print("🟢 Library Quiet")
        print("🔴 RED LED: OFF")

    choice = input("\nCheck again? (y/n): ").lower()
    if choice != 'y':
        print("System Stopped.")
        break
```

```python
# Voice activated calculator - Model Based Agent

import speech_recognition as sr

r = sr.Recognizer()
memory = 0

print("Voice Calculator Started")
print("Say: add 5 3, subtract 10 4, multiply 6 7, divide 20 5")
print("Or say: multiply previous 2")

while True:

    with sr.Microphone() as source:
        print("\nListening...")
        audio = r.listen(source)

    try:
        command = r.recognize_google(audio).lower()
        print("You:", command)

        if command == "exit":
            break

        words = command.split()

        if words[0] == "add":
            memory = float(words[1]) + float(words[2])

        elif words[0] == "subtract":
            memory = float(words[1]) - float(words[2])

        elif words[0] == "multiply":
            if words[1] == "previous":
                memory = memory * float(words[2])
            else:
                memory = float(words[1]) * float(words[2])

        elif words[0] == "divide":
            memory = float(words[1]) / float(words[2])

        print("Answer =", memory)

    except:
        print("Could not understand.")
```

```python
# Smart Parking Assistant - Goal Based Agent

parking_slots = {
    "Slot A": 120,   # distance in meters
    "Slot B": 50,
    "Slot C": 200,
    "Slot D": 80
}

print("🚗 Smart Parking Assistant")
print("--------------------------")

goal = "Find nearest available parking slot"

print("Goal:", goal)

# Find the closest parking slot
nearest_slot = min(parking_slots, key=parking_slots.get)

distance = parking_slots[nearest_slot]

print("\nAvailable Parking Slots:")
for slot, dist in parking_slots.items():
    print(slot, "->", dist, "meters")

print("\n✅ Best Choice:")
print("Park at", nearest_slot)
print("Distance:", distance, "meters")

print("\nNavigation Started...")
print("Follow directions to", nearest_slot)
```

```python
# Smart Energy Management System - Utility Based Agent

print("🏠 Smart Energy Management System")
print("--------------------------------")

# Input conditions
temperature = int(input("Enter room temperature: "))
electricity_price = input("Enter electricity price (high/low): ")
people = input("Are people at home? (yes/no): ")

# Utility scores
comfort_score = 0
cost_score = 0
environment_score = 0

# Comfort evaluation
if temperature > 30 and people == "yes":
    comfort_score = 10     # AC needed
elif temperature > 30:
    comfort_score = 5
else:
    comfort_score = 2


# Electricity cost evaluation
if electricity_price == "high":
    cost_score = -5        # Avoid unnecessary usage
else:
    cost_score = 5


# Environmental impact
if temperature > 30 and people == "yes":
    environment_score = 3
else:
    environment_score = 8


# Calculate total utility
utility = comfort_score + cost_score + environment_score

print("\nUtility Score:", utility)

# Decision making
if utility >= 10:
    print("\nDecision:")
    print("❄️ Turn AC ON at 26°C")
    print("💡 Turn OFF unnecessary lights")

elif utility >= 5:
    print("\nDecision:")
    print("🌬️ Use fan instead of AC")
    print("💡 Reduce lighting usage")

else:
    print("\nDecision:")
    print("✅ No cooling required")
    print("💡 Keep energy usage minimum")
```
