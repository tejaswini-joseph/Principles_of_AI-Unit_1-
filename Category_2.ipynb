```python
# AI Smart Factory Quality Inspector - Fully Observable Environment

print("🏭 AI Smart Factory Quality Inspector")
print("------------------------------------")

# Product information (complete environment state)

product_id = input("Enter Product ID: ")
weight = float(input("Enter product weight (grams): "))
color = input("Enter product color: ")
defect = input("Is there any surface defect? (yes/no): ")

print("\nAnalyzing Product...")

# Quality standards
required_weight = 500
required_color = "blue"

# AI decision making
if (weight >= 480 and weight <= 520
    and color.lower() == required_color
    and defect.lower() == "no"):

    print("\n✅ Product Approved")
    print("Product", product_id, "passed quality inspection.")

else:
    print("\n❌ Product Rejected")
    print("Product", product_id, "failed quality inspection.")

print("\nInspection Completed.")
```

```python
# AI Mars Exploration Rover - Unknown Environment

import random

print("🚀 AI Mars Exploration Rover")
print("----------------------------")

battery = 100
knowledge = []

while battery > 0:

    print("\nExploring new area...")

    # Unknown environment (random discoveries)
    discovery = random.choice([
        "safe path",
        "rock obstacle",
        "scientific sample",
        "dangerous terrain"
    ])

    print("Sensor Detection:", discovery)

    # Learning from environment
    knowledge.append(discovery)

    # Decision making
    if discovery == "rock obstacle":
        print("⚠️ Obstacle detected")
        print("Decision: Changing direction")

    elif discovery == "scientific sample":
        print("🔬 Sample found")
        print("Decision: Collecting sample")

    elif discovery == "dangerous terrain":
        print("⚠️ Unsafe area")
        print("Decision: Moving carefully")

    else:
        print("✅ Safe path found")
        print("Decision: Continue exploration")

    battery -= 20

    print("Battery Remaining:", battery, "%")


print("\n🔋 Battery empty")
print("Returning to base...")
print("\nKnowledge gained:")
print(knowledge)
```

```python
# AI Self-Driving Car - Continuous Environment

import random
import time

print("🚗 AI Self-Driving Car System")
print("-----------------------------")

speed = 50.0

while True:

    # Continuous environment values
    distance = round(random.uniform(5, 50), 2)
    road_curve = round(random.uniform(0, 30), 2)

    print("\nCurrent State:")
    print("Speed:", speed, "km/hr")
    print("Distance from obstacle:", distance, "meters")
    print("Road curve:", road_curve, "degrees")

    print("\nAI Decision:")

    if distance < 10:
        speed -= 5
        print("⚠️ Obstacle close")
        print("Action: Braking")

    elif road_curve > 20:
        speed -= 2
        print("🛣️ Sharp curve detected")
        print("Action: Reducing speed and adjusting steering")

    else:
        speed += 1
        print("✅ Road clear")
        print("Action: Maintaining speed")

    print("New Speed:", round(speed, 2), "km/hr")

    time.sleep(2)
```

```python
# AI Vacuum Cleaning Robot - Single Agent Environment

print("🤖 AI Vacuum Cleaning Robot")
print("----------------------------")

rooms = {
    "Room A": "Clean",
    "Room B": "Dirty",
    "Room C": "Dirty"
}

print("Initial Room Status:")
for room, status in rooms.items():
    print(room, ":", status)

print("\nRobot Starting Cleaning...\n")

# Single agent decision making

for room in rooms:

    if rooms[room] == "Dirty":
        print("🧹 Cleaning", room)
        rooms[room] = "Clean"

    else:
        print("✅", room, "already clean")

print("\nFinal Room Status:")

for room, status in rooms.items():
    print(room, ":", status)

print("\n🎉 Cleaning Completed!")
```

```python
# AI Multi-Agent Traffic System - Multiple agents

print("🚗 AI Multi-Agent Traffic System")
print("--------------------------------")

vehicles = {
    "Car A": {"speed": 50},
    "Car B": {"speed": 70},
    "Ambulance": {"speed": 80}
}

print("Current Vehicle Status:")

for vehicle, data in vehicles.items():
    print(vehicle, "-", data["speed"], "km/hr")


print("\nAI Decisions:")

# Agent interactions

if vehicles["Ambulance"]["speed"] > vehicles["Car A"]["speed"]:
    print("🚑 Ambulance Agent:")
    print("Requesting priority lane")

    print("🚗 Car A Agent:")
    print("Moving aside to give way")

    print("🚙 Car B Agent:")
    print("Reducing speed for safety")

else:
    print("Traffic moving normally")

print("\nAll agents updated their actions.")
```

```python
# AI Parcel Sorting System - Episodic Environment

print("📦 AI Parcel Sorting System")
print("---------------------------")

while True:

    destination = input("\nEnter parcel destination: ")

    if destination.lower() == "chennai":
        print("📍 Place parcel in Chennai Bin")

    elif destination.lower() == "bangalore":
        print("📍 Place parcel in Bangalore Bin")

    else:
        print("📍 Place parcel in Other Cities Bin")

    choice = input("\nSort another parcel? (yes/no): ")

    if choice.lower() != "yes":
        break

print("\nSorting Completed.")
```

```python
# AI Smart Irrigation System - Sequential Environment

print("🌱 AI Smart Irrigation System")
print("-----------------------------")

soil_moisture = 30

for day in range(1, 4):

    print(f"\nDay {day}")
    print("Current Soil Moisture:", soil_moisture, "%")

    if soil_moisture < 40:
        print("💧 Pump ON - Watering plants")
        soil_moisture += 30
    else:
        print("🚫 Pump OFF - Soil moisture is sufficient")
        soil_moisture -= 15

print("\nMonitoring Completed.")
```
