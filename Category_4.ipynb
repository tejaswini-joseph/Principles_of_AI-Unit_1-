```python
#Breadth-First Search (BFS)
from collections import deque

def breadth_first_search(graph, start, goal):
    # Queue stores tuples of (current_node, path_taken)
    queue = deque([(start, [start])])
    visited = set([start])

    while queue:
        current_node, path = queue.popleft()

        if current_node == goal:
            return path

        for neighbor in graph.get(current_node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))

    return None

# Example Usage:
path = breadth_first_search(graph, 'A', 'H')
print(f"BFS Path from A to H: {path}")
# Output: ['A', 'B', 'E', 'H']
```

```python
#Depth-First Search (DFS)
def depth_first_search(graph, start, goal):
    # Stack stores tuples of (current_node, path_taken)
    stack = [(start, [start])]
    visited = set()

    while stack:
        current_node, path = stack.pop()

        if current_node == goal:
            return path

        if current_node not in visited:
            visited.add(current_node)
            for neighbor in graph.get(current_node, []):
                if neighbor not in visited:
                    stack.append((neighbor, path + [neighbor]))

    return None

# Example Usage:
path = depth_first_search(graph, 'A', 'H')
print(f"DFS Path from A to H: {path}")
# Output: ['A', 'C', 'G', ...] (varies depending on neighbor iteration order)
```

```python
#Depth-Limited Search (DLS)
def depth_limited_search(graph, start, goal, limit):

    def recursive_dls(node, goal, depth, path, visited):
        if node == goal:
            return path
        if depth <= 0:
            return None

        visited.add(node)
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                result = recursive_dls(neighbor, goal, depth - 1, path + [neighbor], visited.copy())
                if result is not None:
                    return result
        return None

    return recursive_dls(start, goal, limit, [start], set())

# Example Usage:
# Path exists at depth 3, so depth limit 2 will fail while limit 3 will succeed
path_limit_2 = depth_limited_search(graph, 'A', 'H', limit=2)
path_limit_3 = depth_limited_search(graph, 'A', 'H', limit=3)

print(f"DLS Path (Limit 2): {path_limit_2}")  # Output: None
print(f"DLS Path (Limit 3): {path_limit_3}")  # Output: ['A', 'B', 'E', 'H']
```

```python
#Bidirectional Search
from collections import deque

def bidirectional_search(graph, start, goal):
    if start == goal:
        return [start]

    # Queue and parent trackers for forward search
    q_start = deque([start])
    parent_start = {start: None}

    # Queue and parent trackers for backward search
    q_goal = deque([goal])
    parent_goal = {goal: None}

    while q_start and q_goal:
        # Step forward from start
        if q_start:
            curr_start = q_start.popleft()
            for neighbor in graph.get(curr_start, []):
                if neighbor not in parent_start:
                    parent_start[neighbor] = curr_start
                    q_start.append(neighbor)

                # Intersection check
                if neighbor in parent_goal:
                    return construct_path(neighbor, parent_start, parent_goal)

        # Step backward from goal
        if q_goal:
            curr_goal = q_goal.popleft()
            for neighbor in graph.get(curr_goal, []):
                if neighbor not in parent_goal:
                    parent_goal[neighbor] = curr_goal
                    q_goal.append(neighbor)

                # Intersection check
                if neighbor in parent_start:
                    return construct_path(neighbor, parent_start, parent_goal)

    return None

def construct_path(intersection, parent_start, parent_goal):
    # Reconstruct path from start to intersection
    path_from_start = []
    curr = intersection
    while curr is not None:
        path_from_start.append(curr)
        curr = parent_start[curr]
    path_from_start.reverse()

    # Reconstruct path from intersection to goal
    path_to_goal = []
    curr = parent_goal[intersection]
    while curr is not None:
        path_to_goal.append(curr)
        curr = parent_goal[curr]

    return path_from_start + path_to_goal

# Example Usage:
path = bidirectional_search(graph, 'A', 'H')
print(f"Bidirectional Path from A to H: {path}")
# Output: ['A', 'B', 'E', 'H']
```
