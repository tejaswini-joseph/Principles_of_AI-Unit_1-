```python
#Greedy best-first search
import heapq

def greedy_bfs(start, is_goal, successors, h):
    visited = {start}
    heap = [(h(start), start, [start])]
    while heap:
        _, state, path = heapq.heappop(heap)
        if is_goal(state):
            return path
        for nxt in successors(state):
            if nxt not in visited:
                visited.add(nxt)
                heapq.heappush(heap, (h(nxt), nxt, path + [nxt]))
    return None
```

```python
#A* search
def a_star(start, is_goal, successors, h):
    g = {start: 0}
    heap = [(h(start), start, [start])]
    while heap:
        _, state, path = heapq.heappop(heap)
        if is_goal(state):
            return path
        for nxt in successors(state):
            ng = g[state] + 1
            if nxt not in g or ng < g[nxt]:
                g[nxt] = ng
                heapq.heappush(heap, (ng + h(nxt), nxt, path + [nxt]))
    return None
```

```python
#AO* search
def ao_star(graph, h, node):
    if node not in graph or not graph[node]:
        return h[node]
    best = float('inf')
    for kind, children, cost in graph[node]:
        if kind == 'OR':
            total = cost + ao_star(graph, h, children)
        else:
            total = cost + sum(ao_star(graph, h, c) for c in children)
        best = min(best, total)
    return best

graph = {
    'A': [('OR', 'B', 1), ('AND', ['C', 'D'], 1)],
    'B': [], 'C': [], 'D': []
}
h = {'A': 0, 'B': 5, 'C': 2, 'D': 2}
print(ao_star(graph, h, 'A'))
```

```python
#Memory bounded search
def ida_star(start, is_goal, successors, h):
    limit = h(start)
    path = [start]
    def dfs(g, limit):
        node = path[-1]
        f = g + h(node)
        if f > limit: return f
        if is_goal(node): return True
        m = float('inf')
        for nxt in successors(node):
            if nxt in path: continue
            path.append(nxt)
            t = dfs(g+1, limit)
            if t is True: return True
            m = min(m, t)
            path.pop()
        return m
    while True:
        t = dfs(0, limit)
        if t is True: return path
        if t == float('inf'): return None
        limit = t
```
