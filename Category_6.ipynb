```python
# Minimax Search — Chess
import chess

def eval(board):
    value = {1: 1, 2: 3, 3: 3, 4: 5, 5: 9, 6: 100}
    return sum(len(board.pieces(p, chess.WHITE))*v -
               len(board.pieces(p, chess.BLACK))*v
               for p, v in value.items())

def minimax(board, depth, max_player):
    if depth == 0 or board.is_game_over():
        return eval(board), None

    best = -999 if max_player else 999
    move = None

    for m in board.legal_moves:
        board.push(m)
        score, _ = minimax(board, depth-1, not max_player)
        board.pop()

        if (max_player and score > best) or (not max_player and score < best):
            best, move = score, m

    return best, move

board = chess.Board()
score, move = minimax(board, 2, board.turn == chess.WHITE)

print("Best move:", move)
print("Score:", score)
```

```python
# Alpha-Beta Pruning — Chess
import chess

def eval(board):
    value = {1:1, 2:3, 3:3, 4:5, 5:9, 6:100}
    return sum(len(board.pieces(p, chess.WHITE))*v -
               len(board.pieces(p, chess.BLACK))*v
               for p, v in value.items())

def alphabeta(board, depth, alpha, beta, max_player):
    if depth == 0 or board.is_game_over():
        return eval(board), None

    best = -999 if max_player else 999
    move = None

    for m in board.legal_moves:
        board.push(m)
        score, _ = alphabeta(board, depth-1, alpha, beta, not max_player)
        board.pop()

        if max_player:
            if score > best:
                best, move = score, m
            alpha = max(alpha, best)
        else:
            if score < best:
                best, move = score, m
            beta = min(beta, best)

        if alpha >= beta:
            break

    return best, move

board = chess.Board()
score, move = alphabeta(board, 2, -999, 999, board.turn == chess.WHITE)

print("Best move:", move)
print("Score:", score)
```

```python
# Minimax Search — Tic-Tac-Toe
def winner(b):
    lines = b + list(zip(*b)) + [
        [b[i][i] for i in range(3)],
        [b[i][2-i] for i in range(3)]
    ]

    for x in lines:
        if x[0] != " " and x.count(x[0]) == 3:
            return x[0]
    return None


def minimax(b, max_player):
    w = winner(b)

    if w == "X": return 1
    if w == "O": return -1
    if all(c != " " for r in b for c in r): return 0

    scores = []

    for i in range(3):
        for j in range(3):
            if b[i][j] == " ":
                b[i][j] = "X" if max_player else "O"
                scores.append(minimax(b, not max_player))
                b[i][j] = " "

    return max(scores) if max_player else min(scores)


board = [
    ["X", "O", " "],
    [" ", "X", " "],
    ["O", " ", " "]
]

best = -999
move = None

for i in range(3):
    for j in range(3):
        if board[i][j] == " ":
            board[i][j] = "X"
            score = minimax(board, False)
            board[i][j] = " "

            if score > best:
                best, move = score, (i, j)

print("Best move:", move)
```
