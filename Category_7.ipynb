```python
#Map Coloring — CSP
colors = ["Red", "Green", "Blue"]

map = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "D"],
    "D": ["B", "C"]
}

color = {}

def valid(region, c):
    for n in map[region]:
        if n in color and color[n] == c:
            return False
    return True

def solve():
    if len(color) == len(map):
        return True

    region = next(r for r in map if r not in color)

    for c in colors:
        if valid(region, c):
            color[region] = c

            if solve():
                return True

            del color[region]

    return False

solve()
print(color)
```

```python
#Crypt-Arithmetic Puzzle — CSP
from itertools import permutations

words = "SEND MORE MONEY"
letters = set("SENDMORY")

for p in permutations(range(10), len(letters)):
    d = dict(zip(letters, p))

    if d["S"] == 0 or d["M"] == 0:
        continue

    send = 1000*d["S"] + 100*d["E"] + 10*d["N"] + d["D"]
    more = 1000*d["M"] + 100*d["O"] + 10*d["R"] + d["E"]
    money = 10000*d["M"] + 1000*d["O"] + 100*d["N"] + 10*d["E"] + d["Y"]

    if send + more == money:
        print(d)
        print(send, "+", more, "=", money)
        break
```

```python
#Local Search for CSP
import random

n = 8

def conflicts(board):
    count = 0

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == j - i:
                count += 1

    return count

board = [random.randrange(n) for _ in range(n)]

while conflicts(board) > 0:
    best = board[:]
    best_score = conflicts(board)

    for col in range(n):
        for row in range(n):
            temp = board[:]
            temp[col] = row
            score = conflicts(temp)

            if score < best_score:
                best, best_score = temp, score

    if best_score == conflicts(board):
        board = [random.randrange(n) for _ in range(n)]
    else:
        board = best

print("Solution:", board)
print("Conflicts:", conflicts(board))
```
