```python
#Medical Diagnosis
# Facts
facts = {
    "fever",
    "cough",
    "body_pain"
}

# Rules: conditions -> conclusion
rules = [
    ({"fever", "cough"}, "flu"),
    ({"fever", "body_pain"}, "infection"),
    ({"flu", "body_pain"}, "needs_rest")
]

# Forward Chaining
def forward():
    changed = True

    while changed:
        changed = False
        for conditions, result in rules:
            if conditions <= facts and result not in facts:
                facts.add(result)
                changed = True

forward()

print("Facts:", facts)

#Backward Chaining
def backward(goal, known=set()):
    if goal in known:
        return True

    for conditions, result in rules:
        if result == goal:
            if all(backward(c, known) for c in conditions):
                return True

    return False

print("Is flu?", backward("flu", facts))
print("Needs rest?", backward("needs_rest", facts))
```

```python
#Autonomous Vehicle
# Facts from sensors
facts = {
    "obstacle_ahead",
    "red_light",
    "low_speed"
}

rules = [
    ({"obstacle_ahead"}, "brake"),
    ({"red_light"}, "stop"),
    ({"low_speed", "obstacle_ahead"}, "maintain_distance")
]

# Forward Chaining
def forward():
    changed = True

    while changed:
        changed = False

        for conditions, result in rules:
            if conditions <= facts and result not in facts:
                facts.add(result)
                changed = True

forward()

print("Vehicle actions:", facts)

#Backward Chaining
def backward(goal):
    if goal in facts:
        return True

    for conditions, result in rules:
        if result == goal:
            return all(backward(c) for c in conditions)

    return False

print("Should the vehicle brake?",
      backward("brake"))

print("Should it stop?",
      backward("stop"))
```
