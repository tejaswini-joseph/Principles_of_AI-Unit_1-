```python
# Bayesian Reasoning for Medical Diagnosis

# Prior probability of having a disease
P_Disease = 0.01

# Probability of a positive test
P_Positive_Given_Disease = 0.95

# Probability of a positive test without disease
P_Positive_Given_NoDisease = 0.05

# Bayes' theorem
P_Positive = (
    P_Positive_Given_Disease * P_Disease +
    P_Positive_Given_NoDisease * (1 - P_Disease)
)

P_Disease_Given_Positive = (
    P_Positive_Given_Disease * P_Disease
) / P_Positive

print("Probability of disease:",
      round(P_Disease_Given_Positive * 100, 2), "%")
```

```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

# Define input variables
temperature = ctrl.Antecedent(
    np.arange(0, 101, 1), 'temperature')

vibration = ctrl.Antecedent(
    np.arange(0, 11, 1), 'vibration')

# Define output variable
maintenance = ctrl.Consequent(
    np.arange(0, 101, 1), 'maintenance')

# Membership functions
temperature['low'] = fuzz.trimf(
    temperature.universe, [0, 0, 50])

temperature['medium'] = fuzz.trimf(
    temperature.universe, [30, 55, 80])

temperature['high'] = fuzz.trimf(
    temperature.universe, [60, 100, 100])

vibration['low'] = fuzz.trimf(
    vibration.universe, [0, 0, 4])

vibration['medium'] = fuzz.trimf(
    vibration.universe, [2, 5, 8])

vibration['high'] = fuzz.trimf(
    vibration.universe, [6, 10, 10])

maintenance['low'] = fuzz.trimf(
    maintenance.universe, [0, 0, 40])

maintenance['medium'] = fuzz.trimf(
    maintenance.universe, [20, 50, 80])

maintenance['high'] = fuzz.trimf(
    maintenance.universe, [60, 100, 100])

# Fuzzy IF-THEN rules
rule1 = ctrl.Rule(
    temperature['low'] & vibration['low'],
    maintenance['low'])

rule2 = ctrl.Rule(
    temperature['medium'] | vibration['medium'],
    maintenance['medium'])

rule3 = ctrl.Rule(
    temperature['high'] | vibration['high'],
    maintenance['high'])

# Create fuzzy control system
system = ctrl.ControlSystem(
    [rule1, rule2, rule3])

simulation = ctrl.ControlSystemSimulation(system)

# Input sensor readings
simulation.input['temperature'] = 75
simulation.input['vibration'] = 8

# Calculate output
simulation.compute()

print("Maintenance Priority:",
      round(simulation.output['maintenance'], 2))
```
